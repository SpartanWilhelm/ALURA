```python
import numpy as np
import urllib.request

url = "https://raw.githubusercontent.com/SpartanWilhelm/ALURA/refs/heads/main/Python%20para%20Data%20Science/NumPy_an%C3%A1lise%20num%C3%A9rica%20eficiente%20com%20Python/numpy-dados/apples_ts.csv"

# abre a URL como arquivo
response = urllib.request.urlopen(url)

# carrega os dados ignorando o cabeçalho
dados = np.loadtxt(response, delimiter=",", usecols=np.arange(1,88,1))

dados
```
