Vale ressaltar que:

â = coeficiente angular;
n = número de elementos;
Y = Moscow;
X = datas



```python
Y = Moscow
X = datas
n = np.size(Moscow)
```

```python
(X**2).shape
```

```python
# Fórmula para calcular o coeficiente angular (a)
a = (n*np.sum(X*Y) - np.sum(X)*np.sum(Y))/(n*np.sum(X**2) - np.sum(X)**2)
```

```python
# Fórmula para calcular o coeficiente angular (b)

b = np.mean(Y) - a *np.mean(X)
```

```python
y = a*X + b
```

```python
np.linalg.norm(Moscow-y)
```
